```python
import os
import re
import json
import fitz  # PyMuPDF
import pytesseract
from pdf2image import convert_from_path
import pandas as pd
import torch
from transformers import pipeline

# Metin Temizleme ve Normalizasyon
def normalize_text(text: str) -> str:
    replacements = {
        'İ': 'i', 'I': 'ı', 'Ş': 'ş', 'Ğ': 'ğ', 'Ü': 'ü', 'Ö': 'ö', 'Ç': 'ç',
        'é': 'e', 'è': 'e', 'à': 'a', 'â': 'a', 'ä': 'a', 'ô': 'o', 'ö': 'o', 'ç': 'ç'
    }
    for search, replace in replacements.items():
        text = text.replace(search, replace)
    return text.lower()

# Metin çıkarma (Yöntem bilgisi/fitz-ocr ayrımı kaldırıldı, sadece metin döner)
def extract_first_page(pdf_path: str, min_char_limit: int = 150) -> str:
    doc = fitz.open(pdf_path)
    first_page_text = ""
    if len(doc) > 0:
        first_page_text = doc[0].get_text().strip()

    if len(first_page_text) >= min_char_limit:
        return first_page_text

    try:
        images = convert_from_path(pdf_path, first_page=1, last_page=1)
        ocr_text = pytesseract.image_to_string(images[0], lang='tur+eng')
        return ocr_text
    except Exception:
        return first_page_text

# Doküman tipi tespiti (Regex)
def detect_document_type(raw_text: str) -> str:
    norm_text = normalize_text(raw_text)

    jcr_keywords = ["jcr", "avrasya", "kredi derecelendirme", "derecelendirme notu", "rating"]
    bdr_keywords = ["bagimsiz denetim", "bdr", "denetim raporu", "guney bagimsiz denetim", "pwc", "ey", "kpmg", "deloitte"]

    jcr_score = sum(1 for kw in jcr_keywords if kw in norm_text)
    bdr_score = sum(1 for kw in bdr_keywords if kw in norm_text)

    if jcr_score > bdr_score:
        return "JCR"
    elif bdr_score > jcr_score:
        return "BDR"
    else:
        return "Bilinmiyor / Diğer"

# LLM ile veri çıkarma
def extract_with_llm(text: str, doc_type: str) -> dict:
    # LLM yanıt işleme kısmı
    try:
        # Örnek LLM/Regex parse akışı
        # raw_response = ...
        # json_match = re.search(r"{.*?}", raw_response, re.DOTALL)
        # if json_match:
        #     return json.loads(json_match.group(0))
        pass
    except Exception:
        pass

    return {"musteri_adi": "Bulunamadı", "yil": "Bulunamadı", "bagli_ortaklik": "Bilinmiyor"}

# Ana İşleme ve Excel'e Kaydetme Fonksiyonu (Yöntem sütunu çıkarıldı)
def process_pdf_and_save(pdf_path: str):
    if not pdf_path or not os.path.exists(pdf_path):
        return "Geçersiz dosya yolu"

    text = extract_first_page(pdf_path)
    doc_type = detect_document_type(text)
    extracted = extract_with_llm(text, doc_type)

    comp_name = extracted.get("musteri_adi", "Bulunamadı")
    year = str(extracted.get("yil", "-"))
    subcompany = extracted.get("bagli_ortaklik", "Bilinmiyor")

    output_file = "arayuz_sonuclari.xlsx"
    new_data = pd.DataFrame([{
        "Dosya Adı": os.path.basename(pdf_path),
        "Doküman Tipi": doc_type,
        "Müşteri Adı": comp_name,
        "Yıl": year,
        "Bağlı Ortaklık": subcompany
    }])

    if os.path.exists(output_file):
        existing_df = pd.read_excel(output_file)
        updated_df = pd.concat([existing_df, new_data], ignore_index=True)
        updated_df.to_excel(output_file, index=False)
    else:
        new_data.to_excel(output_file, index=False)

    return f"İşlendi ve kaydedildi: {os.path.basename(pdf_path)}"
```

```python
# 1. Doküman Tipi Tespiti (Regex ile anahtar kelime arama)
def detect_document_type(raw_text: str) -> str:
    norm_text = normalize_text(raw_text)
    jcr_keywords = ["jcr eurasia rating", "corporate credit rating", "rating bildirim formu", "kredi derecelendirme"]
    bdr_keywords = ["bagimsiz denetim", "denetci raporu", "finansal tablolar", "bagimsiz denetim raporu", "bagimsiz denetci"]

    for kw in jcr_keywords:
        if kw in norm_text:
            return "JCR"
    for kw in bdr_keywords:
        if kw in norm_text:
            return "BDR"
    return "Bilinmiyor"

# 2. LLM Modelini Yükleme (Qwen-2.5-1.5B-Instruct)
print("LLM yükleniyor, lütfen bekleyin...")
model_id = "Qwen/Qwen2.5-1.5B-Instruct"
pipe = pipeline(
    "text-generation",
    model=model_id,
    model_kwargs={"torch_dtype": torch.bfloat16},
    device_map="auto"
)
print("LLM başarıyla yüklendi!")

# 3. LLM ile Bilgi Çıkarma Fonksiyonu (Müşteri adı, yıl ve bağlı ortaklık)
def extract_with_llm(text: str, doc_type: str):
    if doc_type == "JCR":
        system_prompt = (
            "Sen bir finansal analistsin. Verilen JCR metninden yalnızca şu bilgileri JSON formatında çıkar:\n"
            "- musteri_adi: Derecelendirilen şirketin tam unvanı (Asla JCR Eurasia Rating yazma).\n"
            "- yil: Raporun yayın tarihi yılı / Publishing Date yılı (Örn: 2024).\n"
            "- bagli_ortaklik: Belgede bağlı ortaklık veya iştirak vurgusu var mı? ('Var' veya 'Yok').\n"
            "Sadece ve sadece geçerli bir JSON objesi döndür, başka hiçbir açıklama yazma."
        )
    else:
        system_prompt = (
            "Sen bir denetim analistisin. Verilen BDR metninden yalnızca şu bilgileri JSON formatında çıkar:\n"
            "- musteri_adi: Denetlenen şirketin unvanı (Asla denetimi yapan bağımsız denetim kuruluşu adını yazma).\n"
            "- yil: Belgenin ait olduğu hesap dönemi yılı (İmza tarihi değil, denetlenen dönemin yılı).\n"
            "- bagli_ortaklik: Şirketin konsolide veya bağlı ortaklık yapısı var mı? ('Var' veya 'Yok').\n"
            "Sadece ve sadece geçerli bir JSON objesi döndür, başka hiçbir açıklama yazma."
        )

    messages = [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": f"Metin:\n{text[:3500]}"}
    ]

    try:
        outputs = pipe(messages, max_new_tokens=150, do_sample=False, temperature=0.1)
        raw_response = outputs[0]["generated_text"][-1]["content"]
        json_match = re.search(r"\{.*?\}", raw_response, re.DOTALL)
        if json_match:
            return json.loads(json_match.group(0))
    except Exception:
        pass

    return {"musteri_adi": "Bulunamadı", "yil": "Bulunamadı", "bagli_ortaklik": "Bilinmiyor"}

# 4. Klasörleri Tarama ve Excel'e Yazma
folders = {
    'BDR': '/content/drive/MyDrive/bank/Bdr',
    'JCR': '/content/drive/MyDrive/bank/Jcr'
}

records = []

for expected_type, folder_path in folders.items():
    if not os.path.exists(folder_path):
        print(f"Uyarı: {folder_path} klasörü bulunamadı!")
        continue

    files = [f for f in os.listdir(folder_path) if f.lower().endswith('.pdf')]
    print(f"\n[{expected_type}] klasöründe {len(files)} adet PDF bulundu, işleniyor...")

    for file in files:
        pdf_path = os.path.join(folder_path, file)

        # Metin çıkar (Yöntem bilgisi kaldırıldı)
        text = extract_first_page(pdf_path)

        # Tip belirle
        doc_type = detect_document_type(text)
        if doc_type == "Bilinmiyor":
            doc_type = expected_type

        # LLM ile bilgileri çek
        extracted = extract_with_llm(text, doc_type)

        records.append({
            "Dosya Adı": file,
            "Doküman Tipi": doc_type,
            "Müşteri Adı": extracted.get("musteri_adi"),
            "Yıl": extracted.get("yil"),
            "Bağlı Ortaklık": extracted.get("bagli_ortaklik")
        })
        print(f"  -> {file} başarıyla işlendi")

# Excel tablosunu oluştur
df = pd.DataFrame(records)
df.to_excel("sonuc_tablosu.xlsx", index=False)

print("\n--- İŞLEM TAMAMLANDI ---")
display(df)
```
